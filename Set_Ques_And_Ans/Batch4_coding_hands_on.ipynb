# 3. Install a Pretrained Segmentation Model and Perform Inference# 
Problem: You need to perform semantic segmentation using a pretrained DeepLabV3+ model on a custom image


```python
import torch
from torchvision.models.segmentation import deeplabv3_resnet50
from PIL import Image
from torchvision import transforms

model = deeplabv3_resnet50(weights = "DEFAULT")
model.eval()

img = Image.open("flower.jpg").convert("RGB")

transform = transforms.Compose([ transforms.Resize((224, 224)),
                               transforms.ToTensor()])

img = transform(img).unsqueeze(0)

while torch.no_grad():
    output = model(img)["out"]

print("segmentation completed")
print(output.shape)
```

# 4. Inference on a Video File Using a Custom Object Detection Model# 
Problem: Run object detection inference on every frame of a video using a PyTorch model you've trained (or a pretrained one). Display bounding boxes


```python
from ultralytics import YOLO
import cv2

model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = cap.read()

    if not ret:
        break

    result = model(frame)
    frame = result[0].plot()

    cv2.imshow("Detection", frame)

    if cv2.waitkey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindow()
```

# 5. Perform Inference on an Unseen Image Using a Saved Model# 
Problem: You’ve trained and saved a deep learning model. Load the model and predict the class of a new input image.


```python
import tensorflow as tf
import numpy as np

model = tf.keras.models.load_model("model.h5")

img = tf.keras.utils.load_img("flower.jpg", target_size = (224, 224))

img = tf.keras.utils.img_to_array(img)
img = img / 255.0
img = np.extend_dims(img, axis = 0)

prediction = model.predict(img)

print("Prediction:", prediction)
```
